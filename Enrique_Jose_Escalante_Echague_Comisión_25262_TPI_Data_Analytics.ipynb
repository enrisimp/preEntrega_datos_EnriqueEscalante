```python
# TODO: Realizar un análisis exploratorio inicial de los DataFrames.
# Sugerencia: usar métodos como .info(), .describe(), .shape y .columns. dtypes.
def eda(df, nombre):
    print(f"\n{'─'*80}")
    print(f"DATASET: {nombre}")
    print(f"{'─'*80}")
    # Información general
    print(f"Dimensiones: {df.shape[0]:,} filas × {df.shape[1]} columnas")
    print("columnas:", list(df.columns))
    # Tipos de datos
    print("\nTipos de Datos:")
    tipo_counts = df.dtypes.value_counts()
    for tipo, count in tipo_counts.items():
        print(f"- {tipo}: {count} columnas")
    print("dtypes:")
    print(df.dtypes)
    # Análisis de nulos
    nulos = df.isna().sum()
    if nulos.sum() > 0:
        print("\nValores Nulos Detectados:")
        nulos_pct = (nulos / len(df) * 100).round(2)
        for col in nulos[nulos > 0].index:
            print(f"- {col}: {nulos[col]}")
    else:
        print("Sin valores nulos")
    # Muestra de datos
    print("\nPrimeras filas:")
    display(df.head(5))
    # Estadísticas descriptivas
    print("\nDescribe (numérico):")
    display(df.describe(include='number'))
    print(f"\n{'─'*80}\n")
```

### 3️⃣ Calidad de los datos

```python
# TODO: Identificar valores nulos y duplicados en los conjuntos de datos.
# Sugerencia: usar .isna().sum() y .duplicated().sum().
# Documentar las observaciones con print() o comentarios.
def calidad(df, nombre, clave=None):
    # Mostrar título descriptivo con el nombre del DF
    print(f"\n{'─'*80}")
    print(f"🔍 CALIDAD: {nombre}")
    print(f"{'─'*80}")
    # Nulos
    print(f"\nValores Nulos:")
    nulos_df = df.isna().sum().to_frame('cantidad')
    display(nulos_df)
    # Duplicados
    dup_completos = df.duplicated(keep=False).sum()
    print(f"\nFilas Duplicadas (completas): {dup_completos} - {dup_completos/len(df)*100:.2f}% del dataset")
    # Duplicados por clave
    if clave and clave in df.columns:
        dup_clave = df[clave].duplicated(keep=False).sum()
        print(f"\nDuplicados en clave '{clave}': {dup_clave}")

        if dup_clave > 0:
            # Mostrar los valores más duplicados
            top_dups = (
                df[df[clave].duplicated(keep=False)][clave]
                .value_counts()
                .head(5)
            )
            print(f"\nLos 5 valores más duplicados:")
            for valor, count in top_dups.items():
                print(f"- {valor}: {count} veces")
    print(f"{'─'*80}\n")
```

### 4️⃣ Limpieza de datos

```python
# TODO: Limpiar el conjunto de datos eliminando duplicados y caracteres no deseados.
# Crear Copias para trabajar
ventas_clean = ventas.copy()
clientes_clean = clientes.copy()
marketing_clean = marketing.copy()
```

```python
# Sugerencia: aplicar .drop_duplicates(), .str.strip(), .str.replace() según el caso.
# Eliminar filas completamente duplicadas
for nombre, df in [('Ventas', ventas_clean), ('Clientes', clientes_clean), ('Marketing', marketing_clean)]:
    antes = len(df)
    if nombre == 'Ventas':
        ventas_clean = ventas_clean.drop_duplicates()
        despues = len(ventas_clean)
    elif nombre == 'Clientes':
        clientes_clean = clientes_clean.drop_duplicates()
        despues = len(clientes_clean)
    else:
        marketing_clean = marketing_clean.drop_duplicates()
        despues = len(marketing_clean)

    eliminados = antes - despues
    if eliminados > 0:
        print(f"- {nombre:12}: {eliminados:>4} duplicados eliminados ({eliminados/antes*100:.2f}%)")
    else:
        print(f"- {nombre:12}: Sin duplicados")
# Documentar el proceso y los resultados.
```

```python
# Normalizar fechas
fechas_convertidas = 0
for df in [ventas_clean, clientes_clean, marketing_clean]:
    for col in df.columns:
        if "fecha" in col.lower():  # detecta columnas con la palabra "fecha"
            df[col] = pd.to_datetime(df[col], errors="coerce", dayfirst=True)
            fechas_convertidas += 1

if fechas_convertidas > 0:
    print(f"{fechas_convertidas} columnas de fecha normalizadas")
else:
    print("No se encontraron columnas de fecha para normalizar")
```

```python
#Normalizar Textos
# Función para limpiar texto en columnas tipo string
def normalizar_texto(df):
    for col in df.select_dtypes(include='object').columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(r'[\u200b\t\r\n]', '', regex=True)
            .str.replace(r'\s+', ' ', regex=True)
            .str.title()
        )
    return df

ventas_clean = normalizar_texto(ventas_clean)
clientes_clean = normalizar_texto(clientes_clean)
marketing_clean = normalizar_texto(marketing_clean)
```

```python
# Normalizar valores numéricos
# precio
if "precio" in ventas_clean.columns:
    ventas_clean["precio"] = (
        ventas_clean["precio"]
        .astype(str)                        # Convierte todo a texto
        .str.replace("$", "", regex=False)  # Elimina el símbolo $
        .str.replace(",", "", regex=False)  # Elimina comas de miles 1,000  1000
        .str.strip()                        # Quita espacios sobrantes
    )
    ventas_clean["precio"] = pd.to_numeric(ventas_clean["precio"], errors="coerce")
# Cantidad
if "cantidad" in ventas_clean.columns:
    ventas_clean["cantidad"] = pd.to_numeric(ventas_clean["cantidad"], errors="coerce").astype("Int64")
```

```python
# Guardar datos limpios como CSV
#print(ventas_clean.head())
#print(clientes_clean.head())
#print(marketing_clean.head())
ventas_clean.info()
ventas_clean.to_csv("ventas_clean.csv", index=False)
clientes_clean.to_csv("clientes_clean.csv", index=False)
marketing_clean.to_csv("marketing_clean.csv", index=False)

print("✅ Archivos guardados: ventas_clean.csv, clientes_clean.csv, marketing_clean.csv")
```

```python
# Reporte
def reporte_limpieza(df_antes, df_despues, nombre):
    """Compara estado antes/después de limpieza"""
    print("\n" + "-"*80)
    print("📊 REPORTE DE LIMPIEZA")
    print("-"*80)
    print(f"\n{nombre}:")
    print(f"   Registros antes:  {len(df_antes):>6,}")
    print(f"   Registros después: {len(df_despues):>6,}")
    print(f"   Eliminados:        {len(df_antes) - len(df_despues):>6,} ({(1-len(df_despues)/len(df_antes))*100:.2f}%)")

    nulos_antes = df_antes.isna().sum().sum()
    nulos_despues = df_despues.isna().sum().sum()
    print(f"   Nulos antes:       {nulos_antes:>6,}")
    print(f"   Nulos eliminados:  {nulos_antes - nulos_despues:>6,}")


    print("\n" + "-"*80)
```

```python
# Detectar columna de producto
def encontrar_columna(df, candidatos):
    for c in df.columns:
        if any(p in c.lower() for p in candidatos):
            return c
    return None

prod_col = encontrar_columna(ventas_clean, ['producto', 'id_producto', 'sku'])

if prod_col is None:
    print("No se encontró columna de producto")
else:
    print(f"Columna detectada: '{prod_col}'")

# Calcular ingreso
print("\nCálculo de Ingreso por Registro")
ventas_perf = ventas_clean.assign(
    ingreso=ventas_clean['precio'] * ventas_clean['cantidad']
)
print(f"- Nueva columna 'ingreso' = precio × cantidad")
print(f"- Ingreso total: ${ventas_perf['ingreso'].sum():,.2f}")

# Agregar por producto
print("\nAgregación por Producto")
resumen_prod = (
    ventas_perf
    .groupby(by=prod_col, dropna=False, as_index=False)
    .agg(
        registros=('ingreso', 'size'),
        unidades=('cantidad', 'sum'),
        precio_promedio=('precio', 'mean'),
        ingreso_total=('ingreso', 'sum')
    )
)
print(f"- {len(resumen_prod)} productos únicos procesados")

# que muestre solo los productos con alto rendimiento. calcular el percentil 80
# y filtrar los productos que superen ese umbral en ventas.
#quantile(0.8)
# Sugerencia: usar .query() o condiciones con operadores lógicos.

# Calcular percentil 80
print("\nCálculo del Percentil 80")
p80_ingreso = resumen_prod['ingreso_total'].quantile(0.80, interpolation='linear')
print(f"- Percentil 80: ${p80_ingreso:,.2f}")
print(f"- El 20% superior de productos genera ≥ ${p80_ingreso:,.2f}")

# Filtrar alto rendimiento
print("\nFiltrado y Clasificación")
ventas_top = (
    resumen_prod
    .query("ingreso_total >= @p80_ingreso", engine='python')
    .sort_values(by=['ingreso_total', 'unidades'], ascending=[False, False], ignore_index=True)
)

print(f"- {len(ventas_top)} productos clasificados como 'Alto Rendimiento'")
print(f"- Representan: {len(ventas_top)/len(resumen_prod)*100:.1f}% del catálogo")

# Calcular participación
total_ingresos = resumen_prod['ingreso_total'].sum()
ingresos_top = ventas_top['ingreso_total'].sum()
print(f"   Generan: ${ingresos_top:,.2f} ({ingresos_top/total_ingresos*100:.1f}% del total)")

# Mostrar resultados
print("PRODUCTOS - ALTO RENDIMIENTO")
print("="*80)
display(ventas_top.head(10))

# Insight clave
print(f"El {len(ventas_top)/len(resumen_prod)*100:.0f}% de productos generan el {ingresos_top/total_ingresos*100:.0f}% de los ingresos totales")
```

### 6️⃣ Agregación

```python
# TODO: Resumir las ventas por categoría de producto y analizar los ingresos generados.
# Sugerencia: usar .groupby() y .agg() para generar métricas como suma y promedio.
```

```python
print("AGREGACIÓN - ANÁLISIS POR CATEGORÍA")
print("="*80)

# Detectar columna categoría
cat_col = encontrar_columna(ventas_clean, ['categoria', 'categoría', 'rubro'])

if cat_col is None:
    print("No se encontró columna de categoría")
else:
    print(f"Columna detectada: '{cat_col}'\n")

# Asegurar columna ingreso
if 'ingreso' not in ventas_clean.columns:
    ventas_cat = ventas_clean.assign(ingreso=ventas_clean['precio'] * ventas_clean['cantidad'])
else:
    ventas_cat = ventas_clean.copy()

# Agregación
print("Agregación de Métricas por Categoría")
resumen_cat = (
    ventas_cat
    .groupby(by=cat_col, dropna=False, as_index=False)
    .agg(
        ventas=('ingreso', 'size'),
        unidades=('cantidad', 'sum'),
        precio_promedio=('precio', 'mean'),
        ingreso_total=('ingreso', 'sum')
    )
    .sort_values(by='ingreso_total', ascending=False, na_position='last', ignore_index=True)
)

# Calcular ticket promedio
resumen_cat['ticket_promedio_por_venta'] = resumen_cat['ingreso_total'] / resumen_cat['ventas']

# Calcular participación
resumen_cat['participacion_%'] = (resumen_cat['ingreso_total'] / resumen_cat['ingreso_total'].sum() * 100).round(2)

print(f"- {len(resumen_cat)} categorías analizadas\n")

# Mostrar resultados

print("RESUMEN POR CATEGORÍA")
display(resumen_cat)

# Análisis e insights
print("\nANÁLISIS DE PARTICIPACIÓN")

for idx, row in resumen_cat.iterrows():
    categoria = row[cat_col]
    participacion = row['participacion_%']
    ticket = row['ticket_promedio_por_venta']

    print(f"{categoria}")
    print(f"- Participación: {participacion}%")
    print(f"- Ticket promedio: ${ticket:,.2f}")
    print(f"- Unidades vendidas: {row['unidades']:,}")

# Recomendaciones
print("\nRECOMENDACIONES ESTRATÉGICAS")
# Identificar categoría con mejor ticket
mejor_ticket = resumen_cat.loc[resumen_cat['ticket_promedio_por_venta'].idxmax()]
print(f"- Mayor ticket promedio: '{mejor_ticket[cat_col]}' (${mejor_ticket['ticket_promedio_por_venta']:,.2f})")
```
