### 3️⃣ 5. Calidad de Datos: Identificar valores nulos y duplicados en los conjuntos de datos. Documentar el estado inicial de los datos.

```python
# TODO: Identificar valores nulos y duplicados en los conjuntos de datos.
# Sugerencia: usar .isna().sum() y .duplicated().sum().
# Documentar las observaciones con print() o comentarios.
def calidad(df, nombre, clave=None):
    # Mostrar título descriptivo con el nombre del DF
    print(f"\n{'─'*80}")
    print(f"🔍 CALIDAD: {nombre}")
    print(f"{'─'*80}")
    # Nulos
    print(f"\nValores Nulos:")
    nulos_df = df.isna().sum().to_frame('cantidad')
    display(nulos_df)
    # Duplicados
    dup_completos = df.duplicated(keep=False).sum()
    print(f"\nFilas Duplicadas (completas): {dup_completos} - {dup_completos/len(df)*100:.2f}% del dataset")
    # Duplicados por clave
    if clave and clave in df.columns:
        dup_clave = df[clave].duplicated(keep=False).sum()
        print(f"\nDuplicados en clave '{clave}': {dup_clave}")

        if dup_clave > 0:
            # Mostrar los valores más duplicados
            top_dups = (
                df[df[clave].duplicated(keep=False)][clave]
                .value_counts()
                .head(5)
            )
            print(f"\nLos 5 valores más duplicados:")
            for valor, count in top_dups.items():
                print(f"- {valor}: {count} veces")
    print(f"{'─'*80}\n")
```

### 4️⃣ 1. Limpieza de Datos: Limpiar el conjunto de datos eliminando duplicados y caracteres no deseados. Documentar el proceso y los resultados.


```python
# TODO: Limpiar el conjunto de datos eliminando duplicados y caracteres no deseados.
# Crear Copias para trabajar
ventas_clean = ventas.copy()
clientes_clean = clientes.copy()
marketing_clean = marketing.copy()
```

```python
# Sugerencia: aplicar .drop_duplicates(), .str.strip(), .str.replace() según el caso.
# Eliminar filas completamente duplicadas
for nombre, df in [('Ventas', ventas_clean), ('Clientes', clientes_clean), ('Marketing', marketing_clean)]:
    antes = len(df)
    if nombre == 'Ventas':
        ventas_clean = ventas_clean.drop_duplicates()
        despues = len(ventas_clean)
    elif nombre == 'Clientes':
        clientes_clean = clientes_clean.drop_duplicates()
        despues = len(clientes_clean)
    else:
        marketing_clean = marketing_clean.drop_duplicates()
        despues = len(marketing_clean)

    eliminados = antes - despues
    if eliminados > 0:
        print(f"- {nombre:12}: {eliminados:>4} duplicados eliminados ({eliminados/antes*100:.2f}%)")
    else:
        print(f"- {nombre:12}: Sin duplicados")
# Documentar el proceso y los resultados.
```

```python
# Normalizar fechas
fechas_convertidas = 0
for df in [ventas_clean, clientes_clean, marketing_clean]:
    for col in df.columns:
        if "fecha" in col.lower():  # detecta columnas con la palabra "fecha"
            df[col] = pd.to_datetime(df[col], errors="coerce", dayfirst=True)
            fechas_convertidas += 1

if fechas_convertidas > 0:
    print(f"{fechas_convertidas} columnas de fecha normalizadas")
else:
    print("No se encontraron columnas de fecha para normalizar")
```

```python
#Normalizar Textos
# Función para limpiar texto en columnas tipo string
def normalizar_texto(df):
    for col in df.select_dtypes(include='object').columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.strip()
            .str.replace(r'[\u200b\t\r\n]', '', regex=True)
            .str.replace(r'\s+', ' ', regex=True)
            .str.title()
        )
    return df

ventas_clean = normalizar_texto(ventas_clean)
clientes_clean = normalizar_texto(clientes_clean)
marketing_clean = normalizar_texto(marketing_clean)
```

```python
# Normalizar valores numéricos
# precio
if "precio" in ventas_clean.columns:
    ventas_clean["precio"] = (
        ventas_clean["precio"]
        .astype(str)                        # Convierte todo a texto
        .str.replace("$", "", regex=False)  # Elimina el símbolo $
        .str.replace(",", "", regex=False)  # Elimina comas de miles 1,000  1000
        .str.strip()                        # Quita espacios sobrantes
    )
    ventas_clean["precio"] = pd.to_numeric(ventas_clean["precio"], errors="coerce")
# Cantidad
if "cantidad" in ventas_clean.columns:
    ventas_clean["cantidad"] = pd.to_numeric(ventas_clean["cantidad"], errors="coerce").astype("Int64")
```

```python
# Guardar datos limpios como CSV
#print(ventas_clean.head())
#print(clientes_clean.head())
#print(marketing_clean.head())
ventas_clean.info()
ventas_clean.to_csv("ventas_clean.csv", index=False)
clientes_clean.to_csv("clientes_clean.csv", index=False)
marketing_clean.to_csv("marketing_clean.csv", index=False)

print("✅ Archivos guardados: ventas_clean.csv, clientes_clean.csv, marketing_clean.csv")
```

### 5️⃣ 2. Transformación de Datos: Aplicar filtros y transformaciones para crear una tabla de ventas que muestre solo los productos con alto rendimiento.

### Objetivo: construir una tabla de rendimiento por producto y quedarnos sólo con los productos de alto rendimiento.

Conceptos clave:

<h3>Transformación de datos:</h3> son operaciones que crean/derivan nuevas columnas (por ejemplo ingreso = precio * cantidad), normalizan formatos (texto/fechas/números) o filtran filas según un criterio.

<h3>Métrica de ingreso:</h3> para ventas, una métrica típica es ingreso por registro = precio * cantidad. Luego podemos agregar por producto (sumar ingresos y unidades) para medir rendimiento total por producto.

<h3>Agregación:</h3> es resumir muchas filas en pocas, aplicando funciones como sum(), mean(), count() agrupando por una clave (ej., producto).
Ej.: “ingreso total por producto” = suma de todos los ingresos de ese producto.

<h3>Percentil:</h3> el percentil 80 (P80) es un valor tal que el 80% de los datos están por debajo o igual a ese valor y el 20% restante por encima.

Si ingreso_total P80 = 120.000, significa que el 80% de los productos tienen ingreso_total ≤ 120.000 y el 20% ≥ 120.000.

<h3>Alto rendimiento:</h3> aquí lo definimos como top 20% de productos según ingreso_total (>= P80). Es un criterio común cuando no hay umbrales de negocio explícitos.
Alternativas válidas: top-K (p. ej. top 50 productos), percentil 75 (P75) o un umbral fijo de negocio (p. ej., “>= $100.000/mes”), o score estandarizado (z-score).

<h3>Plan paso a paso:</h3>

Detectar la columna de producto (tolerando distintos nombres: producto, id_producto, sku, articulo…).

Calcular ingreso por registro = precio * cantidad.

Agregar por producto para obtener métricas (ingreso_total, unidades, precio_promedio, registros).

Calcular P80 con quantile(q=0.80).

Filtrar productos con ingreso_total >= P80.

Ordenar de mayor a menor.

```python
# Detectar columna de producto
def encontrar_columna(df, candidatos):
    for c in df.columns:
        if any(p in c.lower() for p in candidatos):
            return c
    return None

prod_col = encontrar_columna(ventas_clean, ['producto', 'id_producto', 'sku'])

if prod_col is None:
    print("No se encontró columna de producto")
else:
    print(f"Columna detectada: '{prod_col}'")

# Calcular ingreso
print("\nCálculo de Ingreso por Registro")
ventas_perf = ventas_clean.assign(
    ingreso=ventas_clean['precio'] * ventas_clean['cantidad']
)
print(f"- Nueva columna 'ingreso' = precio × cantidad")
print(f"- Ingreso total: ${ventas_perf['ingreso'].sum():,.2f}")

# Agregar por producto
print("\nAgregación por Producto")
resumen_prod = (
    ventas_perf
    .groupby(by=prod_col, dropna=False, as_index=False)
    .agg(
        registros=('ingreso', 'size'),
        unidades=('cantidad', 'sum'),
        precio_promedio=('precio', 'mean'),
        ingreso_total=('ingreso', 'sum')
    )
)
print(f"- {len(resumen_prod)} productos únicos procesados")

# que muestre solo los productos con alto rendimiento. calcular el percentil 80
# y filtrar los productos que superen ese umbral en ventas.
#quantile(0.8)
# Sugerencia: usar .query() o condiciones con operadores lógicos.

# Calcular percentil 80
print("\nCálculo del Percentil 80")
p80_ingreso = resumen_prod['ingreso_total'].quantile(0.80, interpolation='linear')
print(f"- Percentil 80: ${p80_ingreso:,.2f}")
print(f"- El 20% superior de productos genera ≥ ${p80_ingreso:,.2f}")

# Filtrar alto rendimiento
print("\nFiltrado y Clasificación")
ventas_top = (
    resumen_prod
    .query("ingreso_total >= @p80_ingreso", engine='python')
    .sort_values(by=['ingreso_total', 'unidades'], ascending=[False, False], ignore_index=True)
)

print(f"- {len(ventas_top)} productos clasificados como 'Alto Rendimiento'")
print(f"- Representan: {len(ventas_top)/len(resumen_prod)*100:.1f}% del catálogo")

# Calcular participación
total_ingresos = resumen_prod['ingreso_total'].sum()
ingresos_top = ventas_top['ingreso_total'].sum()
print(f"   Generan: ${ingresos_top:,.2f} ({ingresos_top/total_ingresos*100:.1f}% del total)")
```

```python
# Mostrar resultados
print("PRODUCTOS - ALTO RENDIMIENTO")
print("="*80)
print(f"Umbral (percentil 80) de ingreso_total: {float(p80_ingreso):,.2f}")
print("✅ Productos de ALTO RENDIMIENTO (top 20% por ingreso):")
display(ventas_top.head(10))

# Insight clave
print(f"El {len(ventas_top)/len(resumen_prod)*100:.0f}% de productos generan el {ingresos_top/total_ingresos*100:.0f}% de los ingresos totales")
```

### 6️⃣ 3. Agregación: Resumir las ventas por categoría de producto y analizar los ingresos generados.

**Conceptos y plan**
Objetivo: construir un resumen por categoría de producto con métricas útiles (ingreso total, unidades, cantidad de ventas, ticket promedio).

Conceptos clave:

Agregación: operación que resume muchas filas en menos filas, aplicando funciones (sum, mean, count, etc.) después de agrupar por una clave (aquí, la categoría).

Categoría de producto: atributo que agrupa productos similares (ej., “Electrónica”, “Hogar”).

Ticket promedio por venta: ingreso_total / ventas (dónde ventas es el conteo de filas en esa categoría). Indica el importe medio facturado por cada transacción/registro en la categoría.

Nota: esto no es el “precio promedio” del producto; ese ya se calcula con mean sobre precio.

Consideraciones: outliers pueden distorsionar promedios; a veces conviene mirar también la mediana (median).

```python
print("AGREGACIÓN - ANÁLISIS POR CATEGORÍA")
print("="*80)

# Detectar columna categoría
cat_col = encontrar_columna(ventas_clean, ['categoria', 'categoría', 'rubro'])

if cat_col is None:
    print("No se encontró columna de categoría")
else:
    print(f"Columna detectada: '{cat_col}'\n")

# Asegurar columna ingreso
if 'ingreso' not in ventas_clean.columns:
    ventas_cat = ventas_clean.assign(ingreso=ventas_clean['precio'] * ventas_clean['cantidad'])
else:
    ventas_cat = ventas_clean.copy()

# Agregación
print("Agregación de Métricas por Categoría")
resumen_cat = (
    ventas_cat
    .groupby(by=cat_col, dropna=False, as_index=False)
    .agg(
        ventas=('ingreso', 'size'),
        unidades=('cantidad', 'sum'),
        precio_promedio=('precio', 'mean'),
        ingreso_total=('ingreso', 'sum')
    )
    .sort_values(by='ingreso_total', ascending=False, na_position='last', ignore_index=True)
)

# Calcular ticket promedio
resumen_cat['ticket_promedio_por_venta'] = resumen_cat['ingreso_total'] / resumen_cat['ventas']

# Calcular participación
resumen_cat['participacion_%'] = (resumen_cat['ingreso_total'] / resumen_cat['ingreso_total'].sum() * 100).round(2)

print(f"- {len(resumen_cat)} categorías analizadas\n")
```

```python
# Mostrar resultados

print("RESUMEN POR CATEGORÍA")
display(resumen_cat)

# Análisis e insights
print("\nANÁLISIS DE PARTICIPACIÓN")

for idx, row in resumen_cat.iterrows():
    categoria = row[cat_col]
    participacion = row['participacion_%']
    ticket = row['ticket_promedio_por_venta']

    print(f"{categoria}")
    print(f"- Participación: {participacion}%")
    print(f"- Ticket promedio: ${ticket:,.2f}")
    print(f"- Unidades vendidas: {row['unidades']:,}")

# Recomendaciones
print("\nRECOMENDACIONES ESTRATÉGICAS")
# Identificar categoría con mejor ticket
mejor_ticket = resumen_cat.loc[resumen_cat['ticket_promedio_por_venta'].idxmax()]
print(f"- Mayor ticket promedio: '{mejor_ticket[cat_col]}' (${mejor_ticket['ticket_promedio_por_venta']:,.2f})")
```

### 7️⃣ 4. Integración de Datos: Combinar los sets de datos de ventas y marketing para obtener una visión más amplia de las tendencias.

```python
# TODO: Combinar los sets de datos de ventas y marketing para obtener una visión más amplia de las tendencias.
# Sugerencia: usar pd.merge() especificando la clave común entre ambos DataFrames.
# Documentar cualquier observación relevante sobre la combinación de datos.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

dic_dfs = {}

# Preparar df_ventas_marketing y calcular 'precio_total' en ventas_clean
ventas_con_total_price = ventas_clean.copy()
ventas_con_total_price['precio_total'] = ventas_con_total_price['precio'] * ventas_con_total_price['cantidad']

# Combinar ventas_con_total_price with marketing_clean to get 'canal' info
# Note: Merging sales with marketing campaigns on 'producto' might duplicate sales records
# if a product is part of multiple campaigns. This approach assumes that
# Cada ventaeach sale can be associated with all relevant marketing channels.
df_ventas_marketing = pd.merge(
    ventas_con_total_price,
    marketing_clean,
    on='producto',
    how='left',
    suffixes=('_venta', '_marketing') # to differentiate columns like 'fecha' if present
)

# Casos en que el producto no tiene información en marketing
df_ventas_marketing['canal'] = df_ventas_marketing['canal'].fillna('Sin Campaña')
df_ventas_marketing['costo'] = df_ventas_marketing['costo'].fillna(0)

# Guardar el dataframe en dic_dfs
dic_dfs["df_ventas_marketing"] = df_ventas_marketing
```

```python
# Análisis integrado de ventas y marketing por canal y producto

# 1. Análisis por Canal de Venta
ventas_por_canal = (
    dic_dfs["df_ventas_marketing"]
    .groupby('canal', dropna=False)
    .agg(
        total_monto=('precio_total', 'sum'),
        cantidad_transacciones=('precio_total', 'count'),
        ticket_promedio=('precio_total', 'mean')
    )
    .reset_index()
    .sort_values('total_monto', ascending=False)
)

dic_dfs["df_ventas_por_canal"] = ventas_por_canal
print(f"Canales disponibles: {list(ventas_por_canal['canal'])}")
display(ventas_por_canal)

# Visualización: Gráfico de torta para participación por canal
df = dic_dfs["df_ventas_por_canal"]
colors = sns.color_palette("pastel", n_colors=len(df))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Gráfico de torta
ax1.pie(
    df["total_monto"],
    labels=df["canal"],
    autopct="%1.1f%%",
    colors=colors,
    startangle=90,
    wedgeprops={'edgecolor':'white'}
)
ax1.set_title("Participación por Canal (% del Total)")

# Gráfico de barras
ax2.bar(df['canal'], df['total_monto'], color=colors)
ax2.set_title('Ventas Totales por Canal')
ax2.set_xlabel('Canal de Venta')
ax2.set_ylabel('Monto Total Vendido ($)')
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# 2. Análisis por Producto: Ingresos vs Inversión en Marketing
# Calcular ingresos por producto
df_ventas_x_producto = ventas_clean.groupby('producto').agg(
    ingreso_total=('precio', lambda x: (x * ventas_clean.loc[x.index, 'cantidad']).sum())
).sort_values(by='ingreso_total', ascending=False)

# Calcular inversión en marketing por producto
df_marketing_x_producto = (
    marketing_clean
    .groupby('producto')['costo']
    .sum()
    .reset_index()
    .rename(columns={'costo': 'inversion_marketing'})
)

# Combinar ambos datasets
analisis_producto = pd.merge(
    df_ventas_x_producto.reset_index(),
    df_marketing_x_producto,
    on='producto',
    how='left'
).fillna(0)

# Calcular ROI y eficiencia del gasto
analisis_producto['roi_marketing'] = (
    (analisis_producto['ingreso_total'] - analisis_producto['inversion_marketing']) /
    analisis_producto['inversion_marketing'] * 100
).replace([np.inf, -np.inf], 0)

analisis_producto['porcentaje_costo'] = (
    analisis_producto['inversion_marketing'] / analisis_producto['ingreso_total'] * 100
)

print("\n=== Análisis Integrado: Ventas vs Inversión en Marketing ===")
display(analisis_producto.sort_values('roi_marketing', ascending=False))
```

```python
# 3. Visualizaciones del análisis de eficiencia
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Boxplot: distribución del porcentaje de costo
sns.boxplot(
    y=analisis_producto['porcentaje_costo'],
    ax=axes[0, 0],
    color='lightblue'
)
axes[0, 0].set_title('Distribución del % Costo Marketing sobre Ingresos')
axes[0, 0].set_ylabel('% Costo sobre Ingreso')

# Histograma: frecuencia del porcentaje de costo
sns.histplot(
    analisis_producto['porcentaje_costo'],
    kde=True,
    ax=axes[0, 1],
    color='coral'
)
axes[0, 1].set_title('Histograma: % Costo Marketing')
axes[0, 1].set_xlabel('% Costo sobre Ingreso')
axes[0, 1].set_ylabel('Frecuencia')

# Barras: Top productos por ROI
top_roi = analisis_producto.nlargest(10, 'roi_marketing')
axes[1, 0].barh(top_roi['producto'], top_roi['roi_marketing'], color='seagreen')
axes[1, 0].set_title('Top 10 Productos por ROI de Marketing')
axes[1, 0].set_xlabel('ROI (%)')
axes[1, 0].invert_yaxis()

# Scatter: Relación entre inversión y retorno
axes[1, 1].scatter(
    analisis_producto['inversion_marketing'],
    analisis_producto['ingreso_total'],
    alpha=0.6,
    color='mediumpurple'
)
axes[1, 1].set_title('Inversión Marketing vs Ingresos')
axes[1, 1].set_xlabel('Inversión en Marketing ($)')
axes[1, 1].set_ylabel('Ingreso Total ($)')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 1- Estadística Descriptiva: Calcular medidas de tendencia central y dispersión sobre las ventas. Presentar los resultados en un informe.

```python
# ESTADÍSTICA DESCRIPTIVA DE VENTAS
# Usamos el DataFrame resumen_prod que ya contiene el análisis agregado por producto
df = resumen_prod.copy()

print("=" * 60)
print("ANÁLISIS ESTADÍSTICO DESCRIPTIVO - INGRESOS POR PRODUCTO")
print("=" * 60)

# 1. MEDIDAS DE TENDENCIA CENTRAL
# =========================================
media = df['ingreso_total'].mean()
mediana = df['ingreso_total'].median()
moda = df['ingreso_total'].mode()

print("\n MEDIDAS DE TENDENCIA CENTRAL")
print("-" * 40)
print(f"Media (promedio):     ${media:,.2f}")
print(f"Mediana (valor central): ${mediana:,.2f}")
print(f"Moda (más frecuente): ${moda.values[0] if len(moda) > 0 else 'No hay moda única':,.2f}")


# 2. MEDIDAS DE DISPERSIÓN
# =========================================
rango = df['ingreso_total'].max() - df['ingreso_total'].min()
varianza = df['ingreso_total'].var(ddof=1)
desviacion = df['ingreso_total'].std(ddof=1)
coef_variacion = (desviacion / media) * 100

print("\n MEDIDAS DE DISPERSIÓN")
print("-" * 40)
print(f"Rango:                ${rango:,.2f}")
print(f"Varianza:             {varianza:,.2f}")
print(f"Desviación Estándar:  ${desviacion:,.2f}")
print(f"Coeficiente de Variación: {coef_variacion:.2f}%")

# 3. CUARTILES Y RANGO INTERCUARTÍLICO (IQR)
# =========================================
q1 = df['ingreso_total'].quantile(0.25)
q2 = df['ingreso_total'].quantile(0.50)  # Mediana
q3 = df['ingreso_total'].quantile(0.75)
iqr = q3 - q1

print("\n CUARTILES Y RANGO INTERCUARTÍLICO")
print("-" * 40)
print(f"Q1 (25%):             ${q1:,.2f}")
print(f"Q2 (Mediana, 50%):    ${q2:,.2f}")
print(f"Q3 (75%):             ${q3:,.2f}")
print(f"IQR (Q3-Q1):          ${iqr:,.2f}")


# 4. DETECCIÓN DE OUTLIERS
# =========================================
lim_inf = q1 - 1.5 * iqr
lim_sup = q3 + 1.5 * iqr

outliers = df[(df['ingreso_total'] < lim_inf) | (df['ingreso_total'] > lim_sup)]

print("\n DETECCIÓN DE OUTLIERS (Método IQR)")
print("-" * 40)
print(f"Límite Inferior:      ${lim_inf:,.2f}")
print(f"Límite Superior:      ${lim_sup:,.2f}")
print(f"Outliers detectados:  {len(outliers)} productos")

if len(outliers) > 0:
    print("\nProductos con valores atípicos:")
    for idx, row in outliers.iterrows():
        print(f"  • {row['producto']}: ${row['ingreso_total']:,.2f}")
```

```python

# 5. VISUALIZACIONES
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Gráfico 1: Barras de ingresos por producto
df_ordenado = df.sort_values("ingreso_total", ascending=False)
color_celeste = "#386FC7"

barras = axes[0, 0].bar(df_ordenado["producto"], df_ordenado["ingreso_total"], color=color_celeste)
axes[0, 0].set_title("Ingresos Totales por Producto", fontsize=14, weight="bold")
axes[0, 0].set_xlabel("Producto", fontsize=11)
axes[0, 0].set_ylabel("Ingreso Total ($)", fontsize=11)
axes[0, 0].tick_params(axis='x', rotation=90, labelsize=8)

# Agregar valores dentro de las barras
for barra in barras:
    y = barra.get_height()
    axes[0, 0].text(
        barra.get_x() + barra.get_width() / 2,
        y / 2,
        f"${y:,.0f}",
        ha='center', va='center',
        fontsize=7, color="white", rotation=90
    )

# Gráfico 2: Boxplot para visualizar dispersión y outliers
axes[0, 1].boxplot(df['ingreso_total'], vert=True, patch_artist=True,
                   boxprops=dict(facecolor='lightblue'),
                   medianprops=dict(color='red', linewidth=2))
axes[0, 1].set_title("Boxplot - Distribución de Ingresos", fontsize=14, weight="bold")
axes[0, 1].set_ylabel("Ingreso Total ($)", fontsize=11)
axes[0, 1].grid(axis='y', alpha=0.3)

# Agregar líneas de referencia
axes[0, 1].axhline(media, color='green', linestyle='--', linewidth=1, label=f'Media: ${media:,.0f}')
axes[0, 1].axhline(lim_inf, color='orange', linestyle=':', linewidth=1, label=f'Lím. Inf: ${lim_inf:,.0f}')
axes[0, 1].axhline(lim_sup, color='orange', linestyle=':', linewidth=1, label=f'Lím. Sup: ${lim_sup:,.0f}')
axes[0, 1].legend(fontsize=8)

# Gráfico 3: Histograma con curva de densidad
axes[1, 0].hist(df['ingreso_total'], bins=15, color='skyblue', edgecolor='black', alpha=0.7)
axes[1, 0].axvline(media, color='red', linestyle='--', linewidth=2, label=f'Media: ${media:,.0f}')
axes[1, 0].axvline(mediana, color='green', linestyle='-', linewidth=2, label=f'Mediana: ${mediana:,.0f}')
axes[1, 0].set_title("Histograma - Distribución de Ingresos", fontsize=14, weight="bold")
axes[1, 0].set_xlabel("Ingreso Total ($)", fontsize=11)
axes[1, 0].set_ylabel("Frecuencia", fontsize=11)
axes[1, 0].legend()
axes[1, 0].grid(axis='y', alpha=0.3)

# Gráfico 4: Ingresos por categoría
if 'resumen_cat' in locals() or 'resumen_cat' in globals():
    df_cat = resumen_cat.copy()
    df_cat_ordenado = df_cat.sort_values("ingreso_total", ascending=False)
    color_naranja = "#D4722C"

    barras_cat = axes[1, 1].bar(df_cat_ordenado["categoria"],
                                 df_cat_ordenado["ingreso_total"],
                                 color=color_naranja)
    axes[1, 1].set_title("Ingresos Totales por Categoría", fontsize=14, weight="bold")
    axes[1, 1].set_xlabel("Categoría", fontsize=11)
    axes[1, 1].set_ylabel("Ingreso Total ($)", fontsize=11)
    axes[1, 1].tick_params(axis='x', rotation=45)

    # Agregar valores
    for barra in barras_cat:
        y = barra.get_height()
        axes[1, 1].text(
            barra.get_x() + barra.get_width() / 2,
            y / 2,
            f"${y:,.0f}",
            ha='center', va='center',
            fontsize=9, color="white", rotation=90
        )
else:
    axes[1, 1].text(0.5, 0.5, 'No hay datos de categorías disponibles',
                    ha='center', va='center', transform=axes[1, 1].transAxes)
    axes[1, 1].set_title("Ingresos por Categoría")

plt.tight_layout()
plt.show()
```

## 3- Correlación: Evaluar la relación entre diferentes variables (por ejemplo, precio y cantidad vendida) y mostrar los resultados en un gráfico.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats

# Configuración de estilo
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# =========================================
# 1. Gráfico de Dispersión con Línea de Tendencia
# =========================================

x = resumen_prod['precio_promedio']
y = resumen_prod['unidades']

# Calcular la línea de regresión
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
line = slope * x + intercept

# Crear figura
fig, ax = plt.subplots(figsize=(10, 6))

# Dispersión
ax.scatter(x, y, alpha=0.6, s=100, color='steelblue', edgecolors='black', linewidth=0.5)

# Línea de tendencia
ax.plot(x, line, color='red', linewidth=2, label=f'Línea de tendencia (R² = {r_value**2:.3f})')

# Etiquetas y título
ax.set_xlabel('Precio Promedio ($)', fontsize=12, fontweight='bold')
ax.set_ylabel('Unidades Vendidas', fontsize=12, fontweight='bold')
ax.set_title('Correlación: Precio vs Unidades Vendidas', fontsize=14, fontweight='bold', pad=20)
ax.legend(loc='best', fontsize=10)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f" Coeficiente de correlación (Pearson): {r_value:.3f}")
print(f" R² (coeficiente de determinación): {r_value**2:.3f}")
print(f" Ecuación de la recta: y = {slope:.2f}x + {intercept:.2f}")
print(f" P-valor: {p_value:.4f}")
```

```python
# Comparación de períodos

# Dividimos el dataset en dos mitades temporales
fecha_corte = ventas_perf['fecha_venta'].median()

ventas_antes = ventas_perf[ventas_perf['fecha_venta'] <= fecha_corte]['ingreso'].sum()
ventas_despues = ventas_perf[ventas_perf['fecha_venta'] > fecha_corte]['ingreso'].sum()

plt.figure(figsize=(6,4))
plt.bar(['Primer período', 'Segundo período'],
        [ventas_antes, ventas_despues])

plt.title('Comparación de ingresos por período')
plt.ylabel('Ingresos ($)')
plt.tight_layout()
plt.show()
```

```python

# 📅 Ingresos por día de la semana

# Nombres en ingles
ventas_perf['dia_semana'] = ventas_perf['fecha_venta'].dt.day_name()

# Traducir
day_name_mapping = {
    'Monday': 'Lunes',
    'Tuesday': 'Martes',
    'Wednesday': 'Miércoles',
    'Thursday': 'Jueves',
    'Friday': 'Viernes',
    'Saturday': 'Sábado',
    'Sunday': 'Domingo'
}

# Aplicar Cambios
ventas_perf['dia_semana'] = ventas_perf['dia_semana'].map(day_name_mapping)

orden_dias = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']

ingresos_dia_semana = (
    ventas_perf
    .groupby('dia_semana')['ingreso']
    .sum()
    .reindex(orden_dias)
)

plt.figure(figsize=(8,4))
plt.bar(ingresos_dia_semana.index,
        ingresos_dia_semana.fillna(0).values)

plt.title('Ingresos totales por día de la semana')
plt.xlabel('Día')
plt.ylabel('Ingresos ($)')
plt.tight_layout()
plt.show()
```

```python
# Heatmap - Ingresos por Categoría y Mes

ventas_perf['fecha_venta'] = pd.to_datetime(ventas_perf['fecha_venta'], errors='coerce')
ventas_perf['mes'] = ventas_perf['fecha_venta'].dt.month
ventas_perf['mes_nombre'] = ventas_perf['fecha_venta'].dt.strftime('%B')

# Crear tabla pivote
pivot_ingresos = ventas_perf.pivot_table(
    values='ingreso',
    index='categoria',
    columns='mes',
    aggfunc='sum',
    fill_value=0
)

# Nombres de meses en español
meses_esp = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio',
             'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre']
pivot_ingresos.columns = [meses_esp[m-1] for m in pivot_ingresos.columns]

pivot_ingresos = pivot_ingresos.astype(float) # Explicitly convert to float

fig, ax = plt.subplots(figsize=(14, 8))
sns.heatmap(pivot_ingresos,
            annot=True,  # mostrar valores
            fmt='.0f',   # formato sin decimales
            cmap='YlOrRd',  # paleta amarillo-naranja-rojo
            linewidths=0.5,
            linecolor='gray',
            cbar_kws={'label': 'Ingreso Total ($)'},
            ax=ax)

ax.set_title('Mapa de Calor: Ingresos por Categoría y Mes',
             fontsize=18, fontweight='bold', pad=20)
ax.set_xlabel('Mes', fontsize=14, fontweight='bold')
ax.set_ylabel('Categoría', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
